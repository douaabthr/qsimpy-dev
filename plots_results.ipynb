```python
import pandas as pd
import matplotlib.pyplot as plt

# Charger les résultats
df = pd.read_csv(r"D:\Study\Master\master2\semstre3\PFE\tools\qsimpy_dev\qsimpy\results\custom\custom_results.csv")

# ===============================
# 1. Fidelity over tasks
# ===============================
plt.figure()
plt.plot(df["fidelity"])
plt.title("Fidelity over tasks")
plt.xlabel("Task index")
plt.ylabel("Fidelity")
plt.grid()
plt.show()

# ===============================
# 2. Total time (waiting + execution)
# ===============================
df["total_time"] = df["waiting_time"] + df["execution_time"]

plt.figure()
plt.plot(df["total_time"])
plt.title("Total Time per Task")
plt.xlabel("Task index")
plt.ylabel("Time")
plt.grid()
plt.show()

# ===============================
# 3. Rescheduling count
# ===============================
plt.figure()
plt.plot(df["rescheduling_count"])
plt.title("Rescheduling per Task")
plt.xlabel("Task index")
plt.ylabel("Rescheduling count")
plt.grid()
plt.show()

# ===============================
# 4. Fidelity vs Time (important 🔥)
# ===============================
plt.figure()
plt.scatter(df["total_time"], df["fidelity"])
plt.title("Fidelity vs Total Time")
plt.xlabel("Total Time")
plt.ylabel("Fidelity")
plt.grid()
plt.show()

# ===============================
# 5. Moyenne fidelity par round (si tu as ajouté 'round')
# ===============================
if "round" in df.columns:
    fidelity_per_round = df.groupby("round")["fidelity"].mean()

    plt.figure()
    fidelity_per_round.plot()
    plt.title("Average Fidelity per Round")
    plt.xlabel("Round")
    plt.ylabel("Average Fidelity")
    plt.grid()
    plt.show()

# ===============================
# 6. Histogramme fidelity
# ===============================
plt.figure()
plt.hist(df["fidelity"], bins=20)
plt.title("Fidelity Distribution")
plt.xlabel("Fidelity")
plt.ylabel("Frequency")
plt.grid()
plt.show()
```
